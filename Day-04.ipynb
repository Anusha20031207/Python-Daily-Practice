```python
#type of triangle 1.equlateral
s1 = int(input("Enter first side"))
s2 = int(input("Enter second side"))
s3 = int(input ("Enter third side"))
if((s1+s2<s3) or (s2+s3<s1) or (s1+s3<s2)):
    print("not a trianle")
elif(s1==s2==s3):
    print("equilateral trianle")
elif((s1 == s2) or (s2 == s3) or (s3 == s1)):
    print("isosceless trianle")
else:
    print("scalene triangle")
```

```python
'''elictricty bill calculator
An electric company charges first 100 units : $0.5 per unit next 100 units(101-200): $0.75 per unit above 200 units: $1 per unit'''

u = int(input("Enter no of units"))
if(u >0 and u <= 100):
    ta = u * 0.5
elif(u >100 and u <=200):
    ta = (100*0.5)+((u-100)*0.75)
else:
    ta = (100*0.5)+(100*0.75)+((u-200)*1)
print("total amount : ",ta)
```
