```python
#balance
bal = float(input("Enter the balance amount"))
amt = float(input("Enter the amount to withdraw"))
if bal > amt:
    if(amt % 100 == 0):
        print("transaction successful")
    else:
        print("enter the amount in multplies of 100")
else:
    print("insufficient balance")
```

```python
#discount calculator
bill = float(input("enter the bill amount"))
if(bill >= 5000):
    if bill >= 10000:
        print(bill*20/100,"20% discount")
    else:
        print(bill*10/100,"10% discount")
elif bill >= 2000:
    print(bill*5/100,"5% discount")
else:
    print("no discount")
```
